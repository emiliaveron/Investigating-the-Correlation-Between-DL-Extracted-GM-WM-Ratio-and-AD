```python
import os
import pandas as pd

# Path to your source folder
source_folder = "adni_data/MRIs_CLEANED_FILTERED"

# Get all folder names (subject IDs)
subject_ids = [
    name for name in os.listdir(source_folder)
    if os.path.isdir(os.path.join(source_folder, name))
]

# Create DataFrame
df_subjects = pd.DataFrame({"subject_id": sorted(subject_ids)})

# Save to CSV
output_csv = "subject_ids.csv"
df_subjects.to_csv(output_csv, index=False)

print(f"Saved {len(df_subjects)} subject IDs to {output_csv}")
```

```python
import pandas as pd
import os

path = "adni_data/MRIs_IMAGES_FILTERED_NOMASKS"

all_paths = os.listdir(path)
for i in range(len(all_paths)):
    all_paths[i] = os.path.join(path, all_paths[i])

df = pd.DataFrame(all_paths, columns=["subject"])

df.to_csv("subject_paths.csv", index=False)
```
