```python
import pandas as pd

manifest = pd.read_csv("adni_data/MRI_MANIFEST_CLEANED.csv")
diagnosis = pd.read_csv("adni_data/All_Subjects_DXSUM_14Aug2025.csv")
diagnosis.rename(columns={"PTID":"subject_id"}, inplace=True)
subject_ids = pd.read_csv("subject_ids.csv")
```

```python
merged_df = manifest.merge(diagnosis[['subject_id', 'DIAGNOSIS', 'DXAD']], on='subject_id', how='left')
```

```python
present_ids = set(merged_df['subject_id']) & set(subject_ids['subject_id'])
```

```python
# Filter merged_df to only present IDs
filtered_df = merged_df[merged_df['subject_id'].isin(present_ids)]

# Identify rows (assuming PTGENDER: 1 = male, 2 = female)
male_rows_ad = filtered_df[(filtered_df['PTGENDER'] == 1) & (filtered_df['DXAD'] == 1)]
female_rows_ad = filtered_df[(filtered_df['PTGENDER'] == 2) & (filtered_df['DXAD'] == 1)]
male_rows_no_ad = filtered_df[(filtered_df['PTGENDER'] == 1) & (filtered_df['DXAD'] != 1)]
female_rows_no_ad = filtered_df[(filtered_df['PTGENDER'] == 2) & (filtered_df['DXAD'] != 1)]

# Randomly select rows
yes_AD_man = male_rows_ad.sample(n=20, random_state=42).index
no_AD_man = male_rows_no_ad.sample(n=20, random_state=42).index
yes_AD_woman = female_rows_ad.sample(n=20, random_state=42).index
no_AD_woman = female_rows_no_ad.sample(n=20, random_state=42).index

# Combine all sampled rows into one DataFrame
sampled_df = pd.concat([
    male_rows_ad.loc[yes_AD_man],
    male_rows_no_ad.loc[no_AD_man],
    female_rows_ad.loc[yes_AD_woman],
    female_rows_no_ad.loc[no_AD_woman]
])

# Reset the index
sampled_df = sampled_df.reset_index(drop=True)
```

```python
import os
import shutil

# Paths
source_folder = "adni_data/MRIs_CLEANED_FILTERED"
destination_folder = "adni_data/MRI_SAMPLES"

# Target parent folders
alz_folder = os.path.join(destination_folder, "alzheimers")
no_alz_folder = os.path.join(destination_folder, "no_alzheimers")
os.makedirs(alz_folder, exist_ok=True)
os.makedirs(no_alz_folder, exist_ok=True)

# Clean up subject_id format
sampled_df['subject_id'] = sampled_df['subject_id'].astype(str).str.strip()

# Get actual folder names in source directory
available_ids = {name for name in os.listdir(source_folder) if os.path.isdir(os.path.join(source_folder, name))}

# Loop through sampled subjects
for _, row in sampled_df.iterrows():
    subject_id = row['subject_id']
    has_ad = row['DXAD'] == 1

    # Check if folder exists
    if subject_id not in available_ids:
        print(f"Warning: Subject folder {subject_id} does not exist in source")
        continue

    src_subject_folder = os.path.join(source_folder, subject_id)
    parent_folder = alz_folder if has_ad else no_alz_folder
    target_subject_folder = os.path.join(parent_folder, subject_id)

    # Copy the whole folder (overwrite if exists)
    if os.path.exists(target_subject_folder):
        shutil.rmtree(target_subject_folder)  # remove old copy
```
