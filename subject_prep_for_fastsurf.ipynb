```python
import os
import pandas as pd

def generate_unique_subject_ids(df, subject_col='subject'):
    """
    Generates unique subject IDs based on the 'subject' column.
    If a subject ID repeats, appends a count to make it unique.
    
    Args:
        df (pd.DataFrame): Input dataframe with a column containing subject paths.
        subject_col (str): Name of the column with the subject paths.
        
    Returns:
        pd.DataFrame: DataFrame with a new column 'subject_id' containing unique IDs.
    """
    id_counts = {}  # to keep track of occurrences
    subject_ids = []

    for path in df[subject_col]:
        # Extract base ID
        base_id = os.path.basename(path).split('.')[0]
            
        subject_ids.append(base_id)
    
    df['subject_id'] = subject_ids
    return df
```
